```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the anscombe.json dataset
anscombe_df = pd.read_json('/content/sample_data/anscombe.json')

# Display the unique series in the dataset
print("Series available:", anscombe_df['Series'].unique())
display(anscombe_df.head())

# Plot the four different datasets in Anscombe's Quartet
fig, axes = plt.subplots(2, 2, figsize=(12, 10), sharex=True, sharey=True)
fig.suptitle("Anscombe's Quartet", fontsize=16)

series_names = ['I', 'II', 'III', 'IV']
for i, series in enumerate(series_names):
    ax = axes[i // 2, i % 2]
    sub_df = anscombe_df[anscombe_df['Series'] == series].sort_values(by='X')

    # Scatter plot of the data points
    sns.scatterplot(data=sub_df, x='X', y='Y', ax=ax, s=100, color='blue', label='Data')

    # Fit and plot a linear regression line
    sns.regplot(data=sub_df, x='X', y='Y', ax=ax, scatter=False, color='red', label='Regression Line')

    ax.set_title(f'Series {series}')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.legend()
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the MNIST train small dataset
mnist_df = pd.read_csv('/content/sample_data/mnist_train_small.csv', header=None)

# The first column is the label; columns 1 to 784 are pixel values
label = mnist_df.iloc[0, 0]
pixels = mnist_df.iloc[0, 1:].values

# Reshape the 784 pixels into a 28x28 matrix
image_matrix = pixels.reshape(28, 28)

print(f"Label: {label}")
print(f"Reshaped matrix shape: {image_matrix.shape}")

# Plot the reshaped image
plt.figure(figsize=(5, 5))
plt.imshow(image_matrix, cmap='gray')
plt.title(f"MNIST Digit: {label}")
plt.axis('off')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the MNIST train small dataset (labels are in the first column)
mnist_df = pd.read_csv('/content/sample_data/mnist_train_small.csv', header=None)

# Extract labels and count occurrences of each digit class
label_counts = mnist_df[0].value_counts().sort_index()

# Plot the distribution as a pie chart
plt.figure(figsize=(10, 8))
plt.pie(
    label_counts,
    labels=label_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.tab10.colors
)
plt.title('Distribution of Digit Classes in MNIST Train Small Dataset')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the MNIST test dataset without headers
mnist_test_raw = pd.read_csv('/content/sample_data/mnist_test.csv', header=None)

# Select the first column (index 0) which contains the digit labels
labels = mnist_test_raw[0]

# Plot a horizontal boxplot of the labels using seaborn
plt.figure(figsize=(10, 4))
sns.boxplot(x=labels, color='skyblue')
plt.title('Horizontal Boxplot of MNIST Test Digit Labels (Index 0)')
plt.xlabel('Digit Label (0-9)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the built-in flights dataset from seaborn
flights_df = sns.load_dataset('flights')

# Group by year and calculate the total number of passengers per year
annual_passengers = flights_df.groupby('year')['passengers'].sum().reset_index()

# Plot the line chart
plt.figure(figsize=(10, 6))
plt.plot(annual_passengers['year'], annual_passengers['passengers'], marker='*', color='b', linewidth=2, markersize=8)
plt.title('Total Passengers Over Time (Annual)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Passengers', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the built-in tips dataset
tips_df = sns.load_dataset('tips')

# Display the first few rows
display(tips_df.head())

# Plot scatter plot of total_bill vs tip using arrow markers (e.g., '^' for upward arrows)
plt.figure(figsize=(10, 6))
plt.scatter(
    x=tips_df['total_bill'],
    y=tips_df['tip'],
    marker='^',
    color='darkorange',
    alpha=0.8,
    edgecolors='w',
    s=80
)

plt.title('Total Bill vs Tip', fontsize=14)
plt.xlabel('Total Bill ($)', fontsize=12)
plt.ylabel('Tip ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the built-in tips dataset
tips_df = sns.load_dataset('tips')

# Calculate the average total bill for each day (silencing the FutureWarning by setting observed=False)
avg_bill_per_day = tips_df.groupby('day', observed=False)['total_bill'].mean().reset_index()

# Plotting using matplotlib's barh (horizontal bar chart) with light pink bars
plt.figure(figsize=(10, 6))
plt.barh(
    y=avg_bill_per_day['day'],
    width=avg_bill_per_day['total_bill'],
    color='lightpink',
    edgecolor='black'
)

plt.title('Average Total Bill by Day', fontsize=14)
plt.xlabel('Average Total Bill ($)', fontsize=12)
plt.ylabel('Day of the Week', fontsize=12)
plt.grid(True, axis='x', linestyle='--', alpha=0.7)
plt.show()
```
